```python
fruits = pd.DataFrame({'name': ['apple', 'banana', 'mango'],
                        'price': [1, 2, 3]})

#      name    price
#  0   apple   1
#  1   banana  2
#  2   mango   3
```

```python
fruits.name       # dot style, like reading a property
fruits['name']    # bracket style, like a dictionary
```

```python
fruits.iloc[0]       # the whole 1st row (apple)
fruits.iloc[:, 0]     # the whole 1st column (name) -- the ":" means "all rows"
fruits.iloc[0:2]      # 1st and 2nd rows only (apple, banana) -- stop is excluded, same as list slicing you already know
```

```python
fruits.loc[0, 'name']   # row labeled 0, column named 'name' -> 'apple'
```
